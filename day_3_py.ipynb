```python
for i in range(1,6):
  for j in range(5):
    print(i,end="")
  print()
```

```python
for i in range(5,0,-1):
  for j in range(i):
    print("*",end="")
  print()
```

```python
#hollow square
for i in range(1,6):
  for j in range(1,6):
     if(i==1 or i==5 or j==1 or j==4):
      print("*",end="")
     else:
      print(" ",end="")
  print()
```

```python
for i in range(5):
  for j in range(5):
    if( i==j or i+j==5-1):
      print("*",end="")
    else:
      print(" ",end="")
  print()


```
